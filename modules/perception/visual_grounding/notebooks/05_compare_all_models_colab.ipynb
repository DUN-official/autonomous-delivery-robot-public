# 05 Compare All Four Models

Build the final apples-to-apples metrics and charts for Grounding DINO, OWL-ViT, GPT Vision, and GPT-guided OWL-ViT. This notebook reads the saved per-image CSV files created by Notebooks 03 and 04; it does not rerun model inference.

```python
from pathlib import Path
from google.colab import drive

drive.mount('/content/drive')

# Edit this one line if your project folder is elsewhere in Google Drive.
PROJECT_ROOT = Path('/content/drive/MyDrive/autonomous-delivery-robot/modules/perception/visual_grounding')
if not PROJECT_ROOT.exists():
    raise FileNotFoundError(f'Project folder not found: {PROJECT_ROOT}')
print('Project root:', PROJECT_ROOT)
```

## Validate the four result files

```python
RESULT_FILES = {
    'Grounding DINO': PROJECT_ROOT / 'examples/outputs/grounding_dino_results.csv',
    'OWL-ViT': PROJECT_ROOT / 'examples/outputs/owlvit_results.csv',
    'GPT Vision': PROJECT_ROOT / 'examples/outputs/gpt_vision_results.csv',
    'GPT-guided OWL-ViT': PROJECT_ROOT / 'examples/outputs/gpt_guided_owlvit/gpt_guided_owlvit_results.csv',
}
missing = [f'{name}: {path}' for name, path in RESULT_FILES.items() if not path.is_file()]
if missing:
    raise FileNotFoundError('Missing required model results:\n' + '\n'.join(missing))
print('All four model result files are present.')
```

## Generate identical metrics and charts

```python
%pip -q install matplotlib pandas
import subprocess
import sys

report_script = PROJECT_ROOT / 'scripts/build_final_metrics.py'
subprocess.run(
    [sys.executable, str(report_script), '--project-root', str(PROJECT_ROOT)],
    check=True,
)
print('Report generated under:', PROJECT_ROOT / 'results/final')
```

## Display final table and charts

```python
import pandas as pd
from IPython.display import Image, display

FINAL_DIR = PROJECT_ROOT / 'results/final'
metrics = pd.read_csv(FINAL_DIR / 'final_metrics.csv')
display(metrics)
for filename in [
    'mean_iou_by_model.png',
    'success_rates_by_model.png',
    'bbox_quality_distribution.png',
]:
    print(filename)
    display(Image(filename=str(FINAL_DIR / filename)))
```
